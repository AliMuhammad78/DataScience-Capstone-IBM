## Arithmetic Expression Examples

In this section, we will explore **arithmetic expressions** in Python, which include operations like:

- Addition (`+`)
- Subtraction (`-`)
- Multiplication (`*`)
- Division (`/`)
- Modulus (`%`)
- Exponentiation (`**`)
- Floor Division (`//`)

These expressions allow us to perform calculations and manipulate numerical data.


```python
# Multiplication
a = 5
b = 3
product = a * b
print("Product:", product)

# Addition
x = 10
y = 7
sum_result = x + y
print("Sum:", sum_result)
```

```python
# Convert minutes to hours
minutes = 150
hours = minutes / 60
print(f"{minutes} minutes is equal to {hours} hours")
```
